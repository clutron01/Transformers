```python
from transformers import pipeline
```

```python
classifier = pipeline('sentiment-analysis', model = 'tabularisai/multilingual-sentiment-analysis')
```

```python
review = "The earth is going to implode at some point in the future"
result = classifier(review)
print(result)
```

```python
review = "पृथ्वी भविष्य में किसी समय ध्वस्त हो जाएगी।"
result = classifier(review)
print(result)
```
